# BLEU Evaluation

BLEU score measures lexical similarity between generated answers and reference answers. Higher BLEU scores indicate better semantic overlap.

```python
candidate = "RPO is 5 minutes."

reference = "Recovery Point Objective is configured as 5 minutes."

bleu = evaluator.calculate_bleu(
    candidate,
    reference
)

print("BLEU Score:", bleu)
```

# ROUGE-L Evaluation

ROUGE-L measures longest sequence overlap and evaluates answer completeness and retrieval quality.

```python
rouge = evaluator.calculate_rouge_l(
    candidate,
    reference
)

print("ROUGE-L Score:", rouge)
```

# Grounded Faithfulness Evaluation

Faithfulness evaluates whether generated answers are grounded in retrieved document context instead of hallucinated content.

```python
contexts = [
    "Recovery Point Objective (RPO) for this architecture is configured as 5 minutes."
]

faithfulness = evaluator.calculate_faithfulness(
    candidate,
    contexts
)

print("Faithfulness Score:", faithfulness)
```

# Chunking Strategy Comparison

Three chunking strategies were evaluated:

1. Recursive Chunking
2. Sliding Window Chunking
3. Semantic Chunking

The strategies were benchmarked using retrieval quality and grounded faithfulness metrics.

| Chunking Strategy | BLEU & ROUGE Retrieval Quality | Grounded Faithfulness |
|-------------------|-------------------------------|-----------------------|
| Recursive         | 38%                           | 88%                   |
| Sliding Window    | 41%                           | 91%                   |
| Semantic          | 47%                           | 96%                   |

# Hybrid Retrieval Analysis

The system combines:

- BM25 sparse keyword retrieval
- FAISS dense semantic retrieval

Hybrid retrieval improved both lexical recall and semantic matching accuracy.

# Cross Encoder Re-ranking

Cross-encoder reranking using ms-marco-MiniLM-L-6-v2 improved retrieval precision by prioritizing contextually relevant chunks before answer generation.

The feedback system supports future retrieval optimization and iterative system improvement.
